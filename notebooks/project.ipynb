```python
import pdfplumber
```

```python
pdf_path = "../resume.pdf"
```

```python
with pdfplumber.open(pdf_path) as pdf:
    print("PDF loaded successfully")
    print("Total Pages:", len(pdf.pages))
```

```python
with pdfplumber.open(pdf_path) as pdf:
    
    full_text = ""
    
    for page in pdf.pages:
        text = page.extract_text()
        
        if text:
            full_text += text + "\n"
    
print(full_text)
```

```python
cleaned_text = "\n".join(
    line.strip()
    for line in full_text.splitlines()
    if line.strip()
)

print(cleaned_text)
```

```python
lines = cleaned_text.splitlines()

for line in lines:
    print(line)
```

```python
skills_db = [
    "python",
    "java",
    "c",
    "c++",
    "javascript",
    "react",
    "node.js",
    "mongodb",
    "mysql",
    "django",
    "fastapi",
    "docker",
    "aws",
    "html",
    "css",
    "git",
    "github"
]
```

```python
normalized_text = cleaned_text.lower()
print(normalized_text)
```

```python
import re
```

```python
tokens = re.findall(r"\b\w[\w.+#-]*\b", normalized_text)

print(tokens)
```

```python
extracted_skills = []

for skill in skills_db:
    if skill in tokens:
        extracted_skills.append(skill)


print(extracted_skills)
```

```python
extracted_skills = list(set(extracted_skills))

print(extracted_skills)
```

```python
resume_data = {
    "skills": extracted_skills
}

print(resume_data)
```

```python
extracted_skills = sorted(list(set(extracted_skills)))

resume_data = {
    "skills": extracted_skills
}

print(resume_data)
```

```python
import json
```

```python
print(json.dumps(resume_data, indent=4))
```

```python
job_description = """
We are looking for a Python backend developer with experience in
React, MongoDB, Docker, AWS, Git and FastAPI.
The candidate should have knowledge of REST APIs and cloud deployment.
"""
```

```python
print(job_description)
```

```python
normalized_jd = job_description.lower()

print(normalized_jd)
```

```python
jd_tokens = re.findall(r"\b\w[\w.+#-]*\b", normalized_jd)

print(jd_tokens)
```

```python
jd_skills = []

for skill in skills_db:
    if skill in jd_tokens:
        jd_skills.append(skill)

jd_skills = sorted(list(set(jd_skills)))

print(jd_skills)
```

```python
matched_skills = []

for skill in extracted_skills:
    if skill in jd_skills:
        matched_skills.append(skill)

print(matched_skills)
```

```python
missing_skills = []

for skill in jd_skills:
    if skill not in extracted_skills:
        missing_skills.append(skill)

print(missing_skills)
```

```python
analysis_result = {
    "matched_skills": matched_skills,
    "missing_skills": missing_skills
}

print(json.dumps(analysis_result, indent=4))
```

```python
match_score = (len(matched_skills) / len(jd_skills)) * 100

print(f"Match Score: {match_score:.2f}%")
```

```python
analysis_result = {
    "match_score": round(match_score, 2),
    "matched_skills": matched_skills,
    "missing_skills": missing_skills
}

print(json.dumps(analysis_result, indent=4))
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
embedding = model.encode("Python backend developer")

print(embedding)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
resume_skills_text = " ".join(extracted_skills)

jd_skills_text = " ".join(jd_skills)

print(resume_skills_text)
print(jd_skills_text)
```

```python
resume_embedding = model.encode(resume_skills_text)

jd_embedding = model.encode(jd_skills_text)
```

```python
similarity_score = cosine_similarity(
    [resume_embedding],
    [jd_embedding]
)

semantic_match_score = similarity_score[0][0] * 100

print(f"Semantic Match Score: {semantic_match_score:.2f}%")
```

```python
keyword_match_score = match_score

print(f"Keyword Match Score: {keyword_match_score:.2f}%")
```

```python
final_ats_score = (
    keyword_match_score +
    semantic_match_score
) / 2

print(f"Final ATS Score: {final_ats_score:.2f}%")
```

```python
final_analysis = {
    "keyword_match_score": (round(keyword_match_score, 2)),
    "semantic_match_score": float(round(semantic_match_score, 2)),
    "final_ats_score": float(round(final_ats_score, 2)),
    "matched_skills": matched_skills,
    "missing_skills": missing_skills
}

print(json.dumps(final_analysis, indent=4))
```
